```python
import statsmodels
print(statsmodels.__version__)

from statsmodels.tsa.statespace.sarimax import SARIMAX

print("SARIMAX WORKS!")
```

```python
# Model Selection

model = SARIMAX(
    tata_ts_data,
    order=(0, 1, 0),
    seasonal_order=(0, 0, 1, 35),
    trend="t"
)

model_fit = model.fit(disp=False)

print(model_fit.summary())
```

```python
# Fitted Values

fitted_values = model_fit.fittedvalues

plt.figure(figsize=(10, 5))

plt.plot(tata_ts_data, label="Observed")
plt.plot(fitted_values, label="Fitted")

plt.xlabel("Time")
plt.ylabel("Price")
plt.title("Observed vs Fitted Values")
plt.legend()

plt.show()
```

```python
# Forecasting

forecast = model_fit.get_forecast(steps=10)

forecast_mean = forecast.predicted_mean
forecast_ci = forecast.conf_int()

# Create correct future business-day dates
future_dates = pd.date_range(
    start=tata_ts_data.index[-1],
    periods=11,
    freq="B"
)[1:]

# Assign the correct dates to forecast
forecast_mean.index = future_dates
forecast_ci.index = future_dates

# Plot
plt.figure(figsize=(10, 5))

plt.plot(tata_ts_data, label="Observed")
plt.plot(forecast_mean, label="Forecast")

plt.fill_between(
    future_dates,
    forecast_ci.iloc[:, 0],
    forecast_ci.iloc[:, 1],
    alpha=0.2
)

plt.xlabel("Time")
plt.ylabel("Price")
plt.title("Forecast of Tata Motors Stock Price")
plt.legend()

plt.show()
```
